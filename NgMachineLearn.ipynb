# 实现线性回归模型

```python
import numpy as np
import math

train_x = np.array([1.0, 2.0])
train_y = np.array([300.0, 500.0])


def calc_cost(x, y, w, b):
    m = x.shape[0]
    cost = 0

    for i in range(m):
        f_wb = w * x[i] + b
        cost += (f_wb - y[i]) ** 2
    total_cost = 1 / (2 * m) * cost

    return total_cost


def calc_gradient(x, y, w, b):
    m = x.shape[0]
    dJ_dw, dJ_db = 0, 0
    for i in range(m):
        f_wb = w * x[i] + b
        dJ_dw += (f_wb - y[i]) * x[i]
        dJ_db += f_wb - y[i]
    dJ_dw = (1 / m) * dJ_dw
    dJ_db = (1 / m) * dJ_db

    return dJ_dw, dJ_db


def gradient_descent(
    x, y, w_in, b_in, learn_rate, epoch, cost_function, gradient_function
):
    J_history = []
    p_history = []
    w, b = w_in, b_in
    for i in range(epoch):
        dJ_dw, dJ_db = gradient_function(x, y, w, b)
        b = b - learn_rate * dJ_db
        w = w - learn_rate * dJ_dw

        if i < 100000:  # prevent resource exhaustion
            J_history.append(cost_function(x, y, w, b))
            p_history.append([w, b])
        # Print cost every at intervals 10 times or as many iterations if < 10
        if i % math.ceil(epoch / 10) == 0:
            print(
                f"Iteration {i:4}: Cost {J_history[-1]:0.2e} ",
                f"dj_dw: {dJ_dw: 0.3e}, dj_db: {dJ_db: 0.3e}  ",
                f"w: {w: 0.3e}, b:{b: 0.5e}",
            )
    return w, b, J_history, p_history  # return w and J,w history for graphing
```

```python
w_init = 0
b_init = 0
epoch = 10000
learn_rate = 0.1
w_out, b_out, J_history, p_history = gradient_descent(
    x=train_x,
    y=train_y,
    w_in=w_init,
    b_in=b_init,
    learn_rate=learn_rate,
    epoch=epoch,
    cost_function=calc_cost,
    gradient_function=calc_gradient,
)

print(f"(w,b) found by gradient descent: ({w_out:8.4f},{b_out:8.4f})")
```

# 多变量线性回归模型

```python
train_X = np.array(
    [
        [2104, 5, 1, 45],
        [1416, 3, 2, 40],
        [852, 2, 1, 35],
    ]
)
train_y = np.array(
    [
        460,
        231,
        178,
    ]
)
b_init = 785.1811367994083
w_init = np.array([0.39133535, 18.75376741, -53.36032453, -26.42131618])


def calc_cost(X, y, w, b):
    m = X.shape[0]
    # print(f"m = {m}")
    cost = 0

    for i in range(m):
        f_wb = np.dot(w, X[i]) + b
        cost += (f_wb - y[i]) ** 2

    total_cost = 1 / (2 * m) * cost

    return total_cost


def calc_gradient(X, y, w, b):
    m, n = X.shape
    dJ_dw = np.zeros((n,))
    dJ_db = 0.0

    for i in range(m):
        loss_err = (np.dot(w, X[i]) + b) - y[i]

        for j in range(n):
            dJ_dw[j] = dJ_dw[j] + loss_err * X[i, j]

        dJ_db = dJ_db + loss_err

    dJ_dw = dJ_dw / m
    dJ_db = dJ_db / m

    return dJ_dw, dJ_db


def compute_gradient(X, y, w, b):
    """
    Computes the gradient for linear regression
    Args:
      X (ndarray (m,n)): Data, m examples with n features
      y (ndarray (m,)) : target values
      w (ndarray (n,)) : model parameters
      b (scalar)       : model parameter

    Returns:
      dj_dw (ndarray (n,)): The gradient of the cost w.r.t. the parameters w.
      dj_db (scalar):       The gradient of the cost w.r.t. the parameter b.
    """
    m, n = X.shape  # (例子的数量，特征的数量)
    dj_dw = np.zeros((n,))
    dj_db = 0.0

    for i in range(m):
        err = (np.dot(X[i], w) + b) - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + err * X[i, j]
        dj_db = dj_db + err
    dj_dw = dj_dw / m
    dj_db = dj_db / m

    return dj_db, dj_dw


# calc_cost(train_X, train_y, w=w_init, b=b_init)

tmp_dj_dw, tmp_dj_db = calc_gradient(train_X, train_y, w_init, b_init)
# tmp_dj_dw, tmp_dj_db = compute_gradient(train_X, train_y, w_init, b_init)
print(f"dj_db at initial w,b: {tmp_dj_db}")
print(f"dj_dw at initial w,b: \n {tmp_dj_dw}")
```

```python
import copy
import math


def gradient_descent_dot(X, y, w_in, b_in, learn_rate, epoch, cost_func, gradient_func):
    J_history = []
    w = copy.deepcopy(w_in)
    b = b_in
    for i in range(epoch):
        dJ_dw, dJ_db = gradient_func(X, y, w, b)
        w = w - learn_rate * dJ_dw
        b = b - learn_rate * dJ_db
        # 在每次迭代中保存成本J
        if i < 100000:  # prevent resource exhaustion
            J_history.append(cost_func(X, y, w, b))

        # 每隔10次就打印一次成本，如果<10，则打印相同次数的迭代
        if i % math.ceil(epoch / 10) == 0:
            print(f"Iteration {i:4d}: Cost {J_history[-1]:8.2f}   ")

    return w, b, J_history
```

```python
# 初始化参数
initial_w = np.zeros_like(w_init)
initial_b = 0.0
# 一些梯度下降的设置
iterations = 2000
alpha = 0.0000001
# 运行梯度下降法
w_final, b_final, J_hist = gradient_descent_dot(
    train_X,
    train_y,
    initial_w,
    initial_b,
    cost_func=calc_cost,
    gradient_func=calc_gradient,
    learn_rate=alpha,
    epoch=iterations,
)
print(f"b,w found by gradient descent: {b_final:0.2f},{w_final} ")
m, _ = train_X.shape
for i in range(m):
    print(
        f"prediction: {np.dot(train_X[i], w_final) + b_final:0.2f}, target value: {train_y[i]}"
    )
```

# 特征缩放
不同特征值的取值范围过大的时候，会造成等高线分布不均类似椭圆形。当适当调整数值的时候，会发生梯度的*左右横跳*因此，通过缩放特征值，将所有特征值缩放同一个数量级和数值范围内，修复梯度等高线，便于梯度的调节

![alt text](特征缩放的好处.png)

左边为缩放，右边为缩放，当发生梯度下降的时候，方便更快的实现模型的拟合


## 特征缩放的几种方式

1. 直接缩放到0-1
$$
300 <= x_1 <= 2000 \\
0 <= x_2 <= 5 \\

\begin{align}
x_1 &= \frac{x_1^{(i)}}{2000} \\
x_2 &= \frac{x_2^{(i)}}{5}
\end{align}
$$

2. 均值归一化
    1. 先均值
    2. 计算

$$
x_1 = \frac{x_1 - mean(x_1)}{max(x_1) - min(x_1)}
$$

3. **z**分数归一化
    1. 先均值
    2. 计算特征值标准差 $std()$

$$
x_1 = \frac{x_1 - mean(x_1)}{std(x_1)}
$$



```python
import numpy as np

train_X = np.array(
    [
        [2104, 5, 1, 45],
        [1416, 3, 2, 40],
        [852, 2, 1, 35],
    ],
    dtype=np.float32,
)
train_y = np.array(
    [
        460,
        231,
        178,
    ]
)


def _calc_mean(data: np.ndarray):
    return np.mean(data, axis=0)


def _normalization_operation(data, data_mean, max_val, min_val):
    return (data - data_mean) / (max_val - min_val)


def uniformization(data: np.ndarray):
    # 计算每列的均值、最大值、最小值
    mean_data = _calc_mean(data=data)
    max_vals = np.max(data, axis=0)
    min_vals = np.min(data, axis=0)

    print(f"均值 = {mean_data}")
    print(f"最大值 = {max_vals}")
    print(f"最小值 = {min_vals}")

    # 按列对数据进行归一化操作
    m, n = data.shape
    for j in range(n):
        data[:, j] = _normalization_operation(
            data[:, j], mean_data[j], max_vals[j], min_vals[j]
        )

    print("归一化后的数据：")
    print(data)


def _calc_mean_self(data: np.ndarray):
    return np.mean(data.T, axis=1)


def uniformization_self(self_data: np.ndarray):
    mean_data = _calc_mean_self(data=self_data)
    max_vals = np.max(self_data.T, axis=1)
    min_vals = np.min(self_data.T, axis=1)

    print(f"均值 = {mean_data}")
    print(f"最大值 = {max_vals}")
    print(f"最小值 = {min_vals}")
    m, n = self_data.shape
    for i in range(m):
        for j in range(n):
            # print(f"max_val[j] = {max_vals[j]}")
            # print(f"data[i, j] = {data[i, j]}")
            # _normalization_operation(data[i,j])
            self_data[i, j] = _normalization_operation(
                self_data[i, j],
                mean_data[j],
                max_vals[j],
                min_vals[j],
            )
    print("归一化后的数据：")
    print(self_data)


# uniformization(train_X)
print("=" * 50)
uniformization_self(train_X)
```

## Z分数归一化处理

```python
train_X = np.array(
    [[2104, 5, 1, 45], [1416, 3, 2, 40], [852, 2, 1, 35]],
    dtype=np.float32,
)
train_y = np.array([460, 231, 178])


def _calc_mean(data: np.ndarray):
    return np.mean(data, axis=0)


def _normalization_operation(data: np.ndarray, data_mean):
    return (data - data_mean) / np.std(data)


def z_score_uniformization(data: np.ndarray):
    # 计算每列的均值、最大值、最小值
    mean_data = _calc_mean(data=data)
    print(f"均值 = {mean_data}")

    # 按列对数据进行归一化操作
    m, n = data.shape
    for j in range(n):
        data[:, j] = _normalization_operation(
            data[:, j],
            mean_data[j],
        )

    print("归一化后的数据：")
    print(data)
    
z_score_uniformization(train_X)
```

```python
import numpy as np

train_X = np.array(
    [[2104, 5, 1, 45], [1416, 3, 2, 40], [852, 2, 1, 35]],
    dtype=np.float32,
)
train_y = np.array([460, 231, 178])


def _normalization_operation(data: np.ndarray):
    # 直接在此计算标准差
    data_mean = np.mean(data)
    data_std = np.std(data)
    return (data - data_mean) / data_std


def z_score_uniformization(data: np.ndarray):
    # 按列对数据进行Z-Score归一化操作
    m, n = data.shape
    for j in range(n):
        data[:, j] = _normalization_operation(data[:, j])

    print("归一化后的数据：")
    print(data)


z_score_uniformization(train_X)
```
