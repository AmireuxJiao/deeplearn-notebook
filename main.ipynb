```python
import numpy as np
import torch
import torch.nn as nn

np.__version__
```

```python
x = torch.Tensor([[0.0, 0], [0, 1], [1, 0], [1, 1]])
Y = torch.tensor([[0.0], [1], [1], [0]])

net = nn.Sequential(
    nn.Linear(2, 2, bias=True),
    nn.ReLU(),
    nn.Linear(2, 1, bias=True),
    nn.Sigmoid(),
)

loss_function = nn.MSELoss()
# 较低的学习率，需要通过较多的训练次数，才会发生变换
optimizer = torch.optim.SGD(net.parameters(), lr=0.13)

for step in range(5000):
    optimizer.zero_grad()
    out = net(x)
    loss = loss_function(out, Y)
    loss.backward()
    optimizer.step()
    if step % 500 == 0:
        print(
            f"{step:4d}  loss={loss.item():.6f}  pred={out.detach().squeeze().numpy().round(2)}"
        )

with torch.no_grad():
    print("最终结果:", net(x).numpy().round(2))
```
