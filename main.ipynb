```python
import numpy as np
import torch
import torch.nn as nn

np.__version__
```

```python
x = torch.Tensor([[0.0, 0], [0, 1], [1, 0], [1, 1]])
Y = torch.tensor([[0.0], [1], [1], [0]])

net = nn.Sequential(
    nn.Linear(2, 2, bias=True),
    nn.ReLU(),
    nn.Linear(2, 1, bias=True),
    nn.Sigmoid(),
)

loss_function = nn.MSELoss()
# 较低的学习率，需要通过较多的训练次数，才会发生变换
optimizer = torch.optim.SGD(net.parameters(), lr=0.13)

for step in range(5000):
    optimizer.zero_grad()
    out = net(x)
    loss = loss_function(out, Y)
    loss.backward()
    optimizer.step()
    if step % 500 == 0:
        print(
            f"{step:4d}  loss={loss.item():.6f}  pred={out.detach().squeeze().numpy().round(2)}"
        )

with torch.no_grad():
    print("最终结果:", net(x).numpy().round(2))
```

# torch.nn.Sequential
torch.nn.Sequential 是 PyTorch 中的一个容器模块，用来顺序地构建神经网络模型。它允许你将多个层（如全连接层、激活函数、卷积层等）按顺序组合在一起，简化模型定义的代码，减少手动管理层之间连接的麻烦。
- 快速构建网络
- 简单神经网络：对于一些基础的前馈神经网络（如多层感知机，MLP）和标准卷积网络（如卷积神经网络，CNN）

## torch.nn.Sequential 容器特性：
- 顺序执行：Sequential 是逐层执行的。每层的输出会自动作为下一层的输入。
- 适用简单模型：适用于没有复杂结构（如跳跃连接、分支、共享层等）的模型。
- 不支持动态操作：如果网络结构比较复杂（例如条件判断或多个输入），你应该避免使用 Sequential，直接定义一个自定义的 nn.Module 类。

```python
import torch as th

x = th.randn(4, 2, dtype=th.float32)
Y = th.tensor([[0.0], [1], [1], [0]])

net = th.nn.Sequential(
    nn.Linear(2, 2, bias=True), nn.Sigmoid(), nn.Linear(2, 1, bias=True), nn.Sigmoid()
)

loss_function = th.nn.MSELoss()
optimizer = th.optim.SGD(params=net.parameters(), lr=13.0)

for step in range(5000):
    optimizer.zero_grad()
    out = net(x)
    loss = loss_function(out, Y)
    loss.backward()
    optimizer.step()
    if step % 500 == 0:
        print(
            f"{step:4d}  loss={loss.item():.6f}  pred={out.detach().squeeze().numpy().round(2)}"
        )


with torch.no_grad():
    print("最终结果:", net(x).numpy().round(2))
```

# Logging使用方式

logging.basicConfig()函数具有全局唯一性，当其他框架中定义了logging的basicConfig的时候，只能通过创建一个新的日志对象，之后重新设置才行。
> 在 jupyter lab 中多次执行同一单元格，会重复向日志模块增加多个相同的handler，因此，会多次打印相同内容。因此需要采用**if not**完成非重复增加handler

logging总体架构为：创建新的对象，在其中增加自动化执行的handler，handler能够控制转发到那个地方，比如：文件，邮件等各种内容。自动化执行转发。

```python
import logging

train_logger = logging.getLogger()
train_logger.setLevel(logging.DEBUG)

if not train_logger.hasHandlers():
    console_handler = logging.StreamHandler()
    console_handler.setFormatter(
        logging.Formatter("%(asctime)s - %(name)s - %(levelname)s - %(message)s")
    )
    console_handler.setLevel(logging.DEBUG)
    train_logger.addHandler(console_handler)
    
train_logger.debug("hello world")
```
