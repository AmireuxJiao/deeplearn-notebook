# matplotlib基本绘图方法

1. 构建数据
2. plot绘制折线图，bar绘制柱状图，scatter绘制散点图。
3. 多个绘制函数可以同时使用，比如scatter和plot同时使用得到带散点的折线图
4. x/ylabel绘制xy轴说明
5. show为最终绘制图形


```python
import matplotlib.pyplot as plt
import matplotlib

matplotlib.__version__

x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.scatter(x, y, marker='x', c="r")
plt.plot(x,y)
plt.xlabel("x轴坐标")
plt.ylabel("y轴坐标")
plt.title("简单折线图")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = ["A", "B", "C", "D"]
y = [3, 2, 5, 6]

plt.bar(x, y)
# 添加标题和标签
plt.title("简单的柱状图")
plt.xlabel("类别")
plt.ylabel("值")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 2, 3, 4, 5]
y3 = [10, 20, 30, 40, 50]

plt.plot(x, y1, label="Line 1 (y = 2x + 1)", color="r")  # 红色线
plt.plot(x, y2, label="Line 2 (y = x)", color="g")  # 绿色线
plt.plot(x, y3, label="Line 3 (y = 10x)", color="b")  # 蓝色线

plt.legend()
plt.title("多个折线图与图例")
plt.xlabel("x 轴")
plt.ylabel("y 轴")

plt.show()
```
